```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

start_date = '2020-1-1'
end_date = '2026-9-16'

timezone = "America/New_York"

df_h1 = pd.read_csv("./Trading Data/CADCHF_H1.csv",sep=',', parse_dates=['Date'], index_col='Date')
df_h1 = df_h1.loc[start_date : end_date]
df_h1.index = pd.to_datetime(df_h1.index, utc=True).tz_convert(timezone)

gap_high_low_list = []
hour_list = []

for index, row in df_h1.iterrows():
    gap_high_low_list.append(row['High'] - row['Low'])
    hour_list.append(index.time().hour)

df_h1['price_gap'] = gap_high_low_list
df_h1['hour'] = hour_list

# print(in)
result = df_h1.groupby('hour')['price_gap'].mean()


plt.figure(2)
plt.plot(result.index, result.values)

plt.show()
```

```python
from datetime import datetime

import matplotlib.pyplot as plt
import pandas as pd

df_combine = pd.read_csv("./results/COMBINE_DAILY_EQUITY_GBPJPY.csv",sep=',', parse_dates=['Date'], index_col='Date')

print(df_combine)
current_year = None
yearly_equity = 0
yearly_list = []
yearly_equity_list = []
date_format = '%Y-%m-%d %H:%M:%S%z'

for index in range(len(df_combine)):
    row = df_combine.iloc[index]
    date_obj = datetime.strptime(row.name, date_format)
    if current_year is None:
        current_year = date_obj.year
    else:
        if current_year != date_obj.year:
            yearly_list.append(current_year)
            yearly_equity_list.append(df_combine.iloc[index-1].Equity)
            current_year = None
            print(df_combine.iloc[index-1].Equity)

# for index in range(len(df_combine)):
#     row = df_combine.iloc[index]
#     print(row.name)
    # print(row.Name.date())
           

plt.figure(1)
plt.plot(yearly_list, yearly_equity_list)

plt.show()
```
